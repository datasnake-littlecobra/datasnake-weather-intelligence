## Before/after comparison

Confirm the bandpass filter + normalization visibly changed the trace as
expected, before trusting the windowed output.

```python
import matplotlib.pyplot as plt
import numpy as np
import seisbench.data as sbd
from silver_clean import process_trace, load_config

cfg = load_config()
data = sbd.STEAD(sampling_rate=100, cache="../../data/module2_vibration/bronze/stead_cache")
raw = data.get_waveforms(0)
windows = process_trace(raw, cfg["model"]["sampling_rate_hz"], cfg["model"]["window_length_sec"])

fig, axes = plt.subplots(2, 1, figsize=(10, 5))
axes[0].plot(raw[0])
axes[0].set_title("Raw")
axes[1].plot(windows[0][0])
axes[1].set_title("Filtered + normalized, first window")
plt.tight_layout()
plt.show()
print(f"{len(windows)} windows produced from this trace")
```
