## Sanity check: raw waveform plot

Visualize a couple of raw traces before trusting anything downstream —
per the dev doc's "visualize before modeling" step.

```python
import matplotlib.pyplot as plt
import seisbench.data as sbd

data = sbd.STEAD(sampling_rate=100, cache="../../data/module2_vibration/bronze/stead_cache")
trace = data.get_waveforms(0)
plt.figure(figsize=(10, 3))
plt.plot(trace[0])
plt.title("STEAD sample trace 0 (raw, unfiltered)")
plt.xlabel("sample")
plt.ylabel("amplitude")
plt.show()
```
